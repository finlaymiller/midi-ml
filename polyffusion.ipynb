```python
import mido
from mido import MidiFile, MetaMessage


def set_tempo(midi_path: str, bpm: int = 65):
    """
    Reads a MIDI file, sets its tempo to the specified BPM, and saves it back to disk.

    Parameters:
        midi_path (str): Path to the input MIDI file.
        output_path (str): Path to save the modified MIDI file.
        bpm (int, optional): The beats per minute to set. Defaults to 65 BPM.

    Returns:
        None
    """
    # Convert BPM to microseconds per beat
    microseconds_per_beat = int(60_000_000 / bpm)

    # Load the MIDI file
    mid = MidiFile(midi_path)

    for track in mid.tracks:
        # Iterate through track events and replace/set tempo
        for i, msg in enumerate(track):
            if msg.type == "set_tempo":
                track[i] = MetaMessage("set_tempo", tempo=microseconds_per_beat)
                break
        else:
            # If no tempo event exists, insert at the beginning
            track.insert(0, MetaMessage("set_tempo", tempo=microseconds_per_beat))

    # Save the modified MIDI file
    mid.save("polyffusion/exp/tmp.mid")
    return "polyffusion/exp/tmp.mid"


def set_signature(midi_path: str, output_path: str, bpm: int = 65):
    """
    Reads a MIDI file, sets its tempo to the specified BPM, and saves it back to disk.

    Parameters:
        midi_path (str): Path to the input MIDI file.
        output_path (str): Path to save the modified MIDI file.
        bpm (int, optional): The beats per minute to set. Defaults to 65 BPM.

    Returns:
        None
    """
    # Convert BPM to microseconds per beat
    microseconds_per_beat = int(60_000_000 / bpm)

    # Load the MIDI file
    mid = MidiFile(midi_path)

    for track in mid.tracks:
        # Iterate through track events and replace/set tempo
        for i, msg in enumerate(track):
            if msg.type == "time_signature":
                track[i] = MetaMessage(
                    "time_signature",
                    numerator=4,
                    denominator=4,
                    clocks_per_click=36,
                    notated_32nd_notes_per_beat=8,
                    time=0,
                )
        else:
            # If no tempo event exists, insert at the beginning
            track.insert(
                0,
                MetaMessage(
                    "time_signature",
                    numerator=4,
                    denominator=4,
                    clocks_per_click=36,
                    notated_32nd_notes_per_beat=8,
                    time=0,
                ),
            )

    # Save the modified MIDI file
    mid.save(output_path)
    print(f"Updated MIDI file saved to {output_path}")
```

```python
input_midi = "polyffusion/data/16b_for_polyffusion_65.mid"
output_midi = "polyffusion/data/16b_for_polyffusion_65_4-4.mid"
set_signature(input_midi, output_midi, bpm=65)
```
